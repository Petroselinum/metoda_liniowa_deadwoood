```python
import pandas as pd
```

```python
df = pd.read_csv('C:/Users/Budniak Piotr/Documents/Dla_Grzegorza.csv', sep=';', encoding='utf-8')
df.head()
```

```python
df = df[df.gatunek_najliczniejszy_w_1_warstwie == "BK"]
```

```python
df
```

```python
collection = []

for i, group in df.groupby('trakt'):
    if 3 in group['nr_pow'].values:
        collection.append(group.loc[group['nr_pow'] == 3])
    elif 4 in group['nr_pow'].values:
        collection.append(group.loc[group['nr_pow'] == 4])
    elif 2 in group['nr_pow'].values:
        collection.append(group.loc[group['nr_pow'] == 2])
    elif 5 in group['nr_pow'].values:
        collection.append(group.loc[group['nr_pow'] == 5])
    elif 1 in group['nr_pow'].values:
        collection.append(group.loc[group['nr_pow'] == 1])
```

```python
pd.concat(collection, axis=0).to_excel('C:/Users/Budniak Piotr/Documents/Dla_Grzegorza_propozycje_pow_tylko_BK_pan.xlsx', index=False)
```
